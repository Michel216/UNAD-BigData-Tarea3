```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import from_json, col
from pyspark.sql.types import StructType, StructField, StringType, DoubleType

# 1. Crear la sesión de Spark configurada con el paquete conector de Kafka
spark = SparkSession.builder \
    .appName("UNAD-CalidadAire-Streaming") \
    .config("spark.jars.packages", "org.apache.spark:spark-sql-kafka-0-10_2.12:3.3.0") \
    .getOrCreate()

spark.sparkContext.setLogLevel("WARN")

# 2. Definir el esquema estricto para deserializar los datos JSON
schema = StructType([
    StructField("estacion", StringType(), True),
    StructField("variable", StringType(), True),
    StructField("medicion", DoubleType(), True),
    StructField("timestamp", StringType(), True)
])

# 3. Conexión de lectura en flujo continuo desde el broker de Kafka
df_kafka = spark.readStream \
    .format("kafka") \
    .option("kafka.bootstrap.servers", "localhost:9092") \
    .option("subscribe", "calidad_aire_stream") \
    .option("startingOffsets", "latest") \
    .load()

# 4. Deserialización de la columna binaria 'value' a estructura JSON
df_parsed = df_kafka.selectExpr("CAST(value AS STRING) as json_payload") \
    .select(from_json(col("json_payload"), schema).alias("data")) \
    .select("data.*")

# 5. Salida en vivo procesando micro-batches hacia la consola
query = df_parsed.writeStream \
    .outputMode("append") \
    .format("console") \
    .start()
```
