```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, to_timestamp, avg, max, min, count

# 1. Iniciar sesión de Spark
spark = SparkSession.builder \
    .appName("UNAD_Calidad_Aire_Bogota") \
    .getOrCreate()

# 2. Leer el archivo directamente con el lector JSON de Spark activando multiline
df_raw = spark.read.option("multiline", "true").json("calidad_aire.csv")

# 3. Seleccionar y castear las columnas necesarias
df_clean = df_raw.select(
    col("estacion_id"),
    col("nombre_est").alias("estacion"),
    col("msfl_code").alias("variable"),
    col("med_concentracion_estandar").cast("double").alias("concentracion"),
    to_timestamp(col("med_fecha_inicio"), "yyyy-MM-dd HH:mm:ss").alias("fecha_hora"),
    col("municipio"),
    col("latitud").cast("double").alias("latitud"),
    col("longitud").cast("double").alias("longitud")
).filter(col("concentracion").isNotNull())

# 4. Mostrar vista previa de los datos
print("=== VISTA PREVIA BATCH (10 FILAS) ===")
df_clean.show(10, truncate=False)

# 5. Calcular estadísticas agrupadas por Estación y Variable
print("=== ESTADÍSTICAS POR ESTACIÓN Y VARIABLE ===")
df_stats = df_clean.groupBy("estacion", "variable") \
    .agg(
        count("concentracion").alias("total_mediciones"),
        avg("concentracion").alias("promedio"),
        max("concentracion").alias("maximo"),
        min("concentracion").alias("minimo")
    ) \
    .orderBy("estacion", "variable")

df_stats.show(20, truncate=False)
```

```python
# 1. Guardar el resultado del procesamiento Batch en un archivo CSV para el informe
df_stats.coalesce(1).write.mode("overwrite").option("header", "true").csv("resultado_batch_calidad_aire")

# 2. Resumen descriptivo de los datos para la documentación
print("=== RESUMEN EJECUTIVO PARA EL INFORME UNAD ===")
print(f"Total registros analizados: {df_clean.count()}")
print(f"Estaciones detectadas: {df_clean.select('estacion').distinct().count()}")
print(f"Variables monitoreadas: {df_clean.select('variable').distinct().count()}")
```
